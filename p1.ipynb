```python
import numpy as np
import pandas as pd

arr = np.array([5, 10, 15, 20, 25, 30])
print("Array:", arr)
print("Mean:", arr.mean())
print("Reshaped array:", arr.reshape(3,2))
print("Sliced (1 to 3)", arr[1:5])

data = { 'Name' : ['Ganesh', 'Ravi', 'Hruthik','Goutham'],
         'Score' : [90, 56, 89, 67],
         'Branch' : ['CSE','MECH','ISE','AIML']
       }

df = pd.DataFrame(data)
print(df)

print("\nFirst 1 rows:\n", df.iloc[0:1])
print("\nMarks Column:\n", df["Score"])
print("\nRows where marks > 60\n", df[df['Score'] > 60])
```
